```python
from collections import Counter
from itertools import combinations

genre_pairs = Counter()

for genre_string in df["Genres"].dropna():
    game_genres = sorted(set(
        genre.strip()
        for genre in genre_string.split(",")
        if genre.strip()
    ))

    genre_pairs.update(combinations(game_genres, 2))

genre_pair_counts = pd.DataFrame(
    [
        {"Genre 1": pair[0], "Genre 2": pair[1], "Count": count}
        for pair, count in genre_pairs.most_common(20)
    ]
)

display(genre_pair_counts)
```

```python
tag_pairs = Counter()

for tag_string in df["Tags"].dropna():
    game_tags = sorted(set(
        tag.strip()
        for tag in tag_string.split(",")
        if tag.strip()
    ))

    tag_pairs.update(combinations(game_tags, 2))

tag_pair_counts = pd.DataFrame(
    [
        {"Tag 1": pair[0], "Tag 2": pair[1], "Count": count}
        for pair, count in tag_pairs.most_common(20)
    ]
)

display(tag_pair_counts)
```

```python
tag_pair_counts["Pair"] = (
    tag_pair_counts["Tag 1"] + " + " + tag_pair_counts["Tag 2"]
)

tag_pair_counts.sort_values("Count").plot(
    x="Pair",
    y="Count",
    kind="barh",
    figsize=(12, 8),
    legend=False,
    title="20 Most Common Steam Tag Combinations"
)

plt.xlabel("Number of Games")
plt.ylabel("Tag Combination")
plt.show()
```

```python
# Number of games with at least one tag
N = df["Tags"].notna().sum()

# Individual tag frequencies
tag_frequency = tag_counts.to_dict()

# Calculate lift for every observed tag pair
lift_results = []

for (tag1, tag2), pair_count in tag_pairs.items():
    count1 = tag_frequency[tag1]
    count2 = tag_frequency[tag2]

    lift = (pair_count * N) / (count1 * count2)

    lift_results.append({
        "Tag 1": tag1,
        "Tag 2": tag2,
        "Count": pair_count,
        "Lift": lift
    })
```

```python
lift_df = pd.DataFrame(lift_results)

display(lift_df.head(20))
```

```python
reference_date = pd.Timestamp("2026-11-01")

eligibility = pd.DataFrame({
    "Has name": df["Name"].notna() & df["Name"].str.strip().ne(""),
    "Has genre or tag": (
        df["Genres"].fillna("").str.strip().ne("") |
        df["Tags"].fillna("").str.strip().ne("")
    ),
    "Released": df["Release date"].le(reference_date),
    "Has reviews": df["Total reviews"] > 0
})

eligibility_summary = pd.DataFrame({
    "Games": eligibility.sum(),
    "Percentage": (eligibility.mean() * 100).round(2)
})

display(eligibility_summary)
```

```python
basic_candidates = (
    eligibility["Has name"] &
    eligibility["Has genre or tag"] &
    eligibility["Released"]
)

print("Total games:", len(df))
print("Basic recommendation candidates:", basic_candidates.sum())
print("Excluded games:", (~basic_candidates).sum())
print("Candidate coverage:", round(basic_candidates.mean() * 100, 2), "%")
```

```python
exclusion_reasons = pd.DataFrame({
    "Missing name": ~eligibility["Has name"],
    "Missing genre and tag": ~eligibility["Has genre or tag"],
    "Not released": ~eligibility["Released"]
})

display(exclusion_reasons.sum())

display(
    df.loc[
        ~basic_candidates &
        eligibility["Has genre or tag"],
        ["AppID", "Name", "Release date", "Genres", "Tags"]
    ]
)
```
